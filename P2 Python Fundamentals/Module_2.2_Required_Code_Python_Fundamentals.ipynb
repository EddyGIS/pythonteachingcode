## Program: list-o-matic  
This program takes string input and checks if that string is in a list of strings    
- if string is in the list it removes the first instance from list  
- if string is not in the list the input gets appended to the list  
- if the string is empty then the last item is popped from the list 
- if the **list becomes empty** the program ends  
- if the user enters "quit" then the program ends  

program has 2 parts  
- **program flow** which can be modified to ask for a specific type of item.  This is the programmers choice.  Add a list of fish, trees, books, movies, songs.... your choice.

````python
# P2M2 List-o-matic
# Edward Graham

def list_o_matic(location, locations):
    if location == "":
        removed = locations.pop()
        return removed + " popped from list"

    elif location in locations:
        locations.remove(location)
        return "1 instance of " + location + " removed from list"

    else:
        locations.append(location)
        return "1 instance of " + location + " appended to list"


# Starting list of Alaska locations
locations = ["Bettles", "Tanana", "Kaktovik", "Shishmaref"]

while locations:
    print("\nWelcome, Edward Graham. Look at all the Alaska locations", locations)

    location = input("Enter an Alaska location (or quit): ")

    if location.lower() == "quit":
        break

    message = list_o_matic(location, locations)
    print(message)

print("Goodbye!")
````
